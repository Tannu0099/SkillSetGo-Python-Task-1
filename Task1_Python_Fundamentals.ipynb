## 2. Control Flow
`if / elif / else`, `for`, `while`, and comprehensions.

```python
def grade_from_score(score):
    """Return a letter grade for a numeric score (0-100)."""
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F"

scores = [95, 82, 67, 40, 100]
for s in scores:
    print(f"Score {s:3d} -> Grade {grade_from_score(s)}")
```

```python
# while loop: simple countdown / simulation of an epoch counter
epoch = 5
while epoch > 0:
    print(f"Training epoch {epoch}...")
    epoch -= 1
print("Training complete.")
```

## 3. Functions
Regular functions, default args, `*args` / `**kwargs`, and `lambda`.

```python
def normalize(values, min_val=0, max_val=1):
    """Min-max normalize a list of numbers into [min_val, max_val]."""
    lo, hi = min(values), max(values)
    return [min_val + (v - lo) * (max_val - min_val) / (hi - lo) for v in values]

raw = [10, 20, 30, 40, 50]
print("Raw:       ", raw)
print("Normalized:", [round(v, 2) for v in normalize(raw)])
```

```python
def describe_model(name, *metrics, **hyperparams):
    """Demonstrates *args and **kwargs."""
    print(f"Model: {name}")
    print(f"  Metrics reported: {metrics}")
    print(f"  Hyperparameters: {hyperparams}")

describe_model("LogisticRegression", "accuracy", "f1_score", C=1.0, max_iter=200)

square = lambda x: x ** 2
print("Lambda square(9):", square(9))
```

## 4. Data Structures
Lists, tuples, dictionaries, sets, and comprehensions — the everyday tools for handling data before it becomes a DataFrame.

```python
students = [
    {"name": "Aarav", "score": 88, "track": "AI/ML"},
    {"name": "Diya",  "score": 92, "track": "AI/ML"},
    {"name": "Kabir", "score": 75, "track": "Web Dev"},
]

# List comprehension: names of students scoring above 80
top_students = [s["name"] for s in students if s["score"] > 80]
print("Top students:", top_students)

# Dict comprehension: name -> score map
score_map = {s["name"]: s["score"] for s in students}
print("Score map:", score_map)

# Set: unique tracks
tracks = {s["track"] for s in students}
print("Unique tracks:", tracks)

# Tuple: immutable record
point = (10, 20)
print("Tuple point:", point)
```

## 5. String Manipulation
Common operations used when cleaning text data.

```python
raw_text = "   Skill Set Go -- AI/ML Internship 2026!!!  "

cleaned = raw_text.strip()
lowered = cleaned.lower()
no_punct = "".join(ch for ch in lowered if ch.isalnum() or ch.isspace())
words = no_punct.split()

print("Original :", repr(raw_text))
print("Cleaned  :", repr(cleaned))
print("No punct :", no_punct)
print("Word list:", words)
print("Word count:", len(words))
```

## 6. File Handling
Writing and reading a small CSV file without external libraries.

```python
import csv

rows = [
    {"name": "Aarav", "score": 88},
    {"name": "Diya", "score": 92},
    {"name": "Kabir", "score": 75},
]

with open("students.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerows(rows)

print("students.csv written.")

with open("students.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
```

## 7. Object-Oriented Programming
A small class to model a dataset of student scores — encapsulating data + behavior, a pattern used later for model wrapper classes.

```python
class ScoreBook:
    """A tiny class that stores student scores and computes basic stats."""

    def __init__(self, records):
        self.records = records  # list of dicts: {"name": str, "score": float}

    def average(self):
        scores = [r["score"] for r in self.records]
        return sum(scores) / len(scores) if scores else 0

    def top_scorer(self):
        return max(self.records, key=lambda r: r["score"])

    def __len__(self):
        return len(self.records)

    def __repr__(self):
        return f"ScoreBook({len(self)} records, avg={self.average():.2f})"


book = ScoreBook(students)  # 'students' defined earlier
print(book)
print("Average score:", round(book.average(), 2))
print("Top scorer   :", book.top_scorer())
```
